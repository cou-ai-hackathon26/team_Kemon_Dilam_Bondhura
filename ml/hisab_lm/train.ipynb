# Hisab-LM training
Open this notebook in a free CUDA runtime with a checkout of Capzoryx. Run from the repository root. Obtain the synthetic ZIP from the project. Only synthetic training and validation splits are loaded.

```python
import os, pathlib, subprocess, sys
root=pathlib.Path.cwd()
while root!=root.parent and not (root/"ml/hisab_lm/train.py").exists(): root=root.parent
assert (root/"ml/hisab_lm/train.py").exists(), "Start with a checkout of the Capzoryx repository"
os.chdir(root)
subprocess.run([sys.executable,"-m","pip","install","-r","ml/hisab_lm/requirements-gpu.lock.txt"],check=True)
```

Restart the notebook kernel if the provider preloaded a different torch version. Set the archive path and selected base model. The default below is a small starting configuration. Record bake-off measurements and licensing before choosing the production base.

```python
import os, pathlib, subprocess, sys
from huggingface_hub import model_info
base="Qwen/Qwen3-1.7B"
revision=model_info(base).sha
archive=pathlib.Path("web/public/datasets/hisab-lm-synthetic-v1.zip").resolve()
assert archive.exists(), "Set archive to the supplied synthetic ZIP"
print(base,revision)
```

```python
subprocess.run([sys.executable,"-m","ml.hisab_lm.train","--archive",str(archive),"--base",base,"--revision",revision,"--output",".private/hisab-lm","--gguf"],check=True)
```

Training writes the adapter, validation evidence and Q4_K_M GGUF locally. Store the selected model in private storage with its base license and notices. Follow MODEL_CARD.md to run inference and compare results before enabling individual tasks.
